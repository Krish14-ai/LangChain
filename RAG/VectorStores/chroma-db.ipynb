```python
from langchain_ollama.embeddings import OllamaEmbeddings
from langchain_chroma import Chroma
from langchain_classic.schema import Document
from langchain_community.document_loaders import PyPDFLoader
```

```python
embeddings = OllamaEmbeddings(
    model = 'nomic-embed-text'
)
```

```python
path= r'C:\Users\Krish\Downloads\LangChain\RAG\Books\Agentic-Artificial-Intelligence.pdf'

book = PyPDFLoader(file_path= path)
```

```python
page1 = Document(
    page_content= book.load_and_split()[10].page_content,
    metadata = book.load_and_split()[10].metadata
)

page2 = Document(
    page_content= book.load_and_split()[11].page_content,
    metadata = book.load_and_split()[11].metadata
)

page3=  Document(
    page_content= book.load_and_split()[12].page_content,
    metadata = book.load_and_split()[12].metadata 
)
```

```python
pages = [page1,page2,page3]
```

```python
vector_store = Chroma(
    embedding_function= embeddings,
    persist_directory= 'chroma_db',
    collection_name = 'sample'
)
```

```python
vector_store.add_documents(pages)
```
