```python
from langchain_ollama.embeddings import OllamaEmbeddings
from langchain_chroma import Chroma
from langchain_classic.schema import Document
from langchain_community.document_loaders import PyPDFLoader
```

```python
embeddings = OllamaEmbeddings(
    model = 'nomic-embed-text'
)
```

```python
path= r'C:\Users\Krish\Downloads\LangChain\RAG\Books\Agentic-Artificial-Intelligence.pdf'

book = PyPDFLoader(file_path= path)
```

```python
page1 = Document(
    page_content= book.load_and_split()[14].page_content,
    metadata = book.load_and_split()[14].metadata
)

page2 = Document(
    page_content= book.load_and_split()[15].page_content,
    metadata = book.load_and_split()[15].metadata
)

page3=  Document(
    page_content= book.load_and_split()[16].page_content,
    metadata = book.load_and_split()[16].metadata 
)
```

```python
pages = [page1,page2,page3]
```

```python
vector_store = Chroma(
    embedding_function= embeddings,
    persist_directory= 'chroma_db',
    collection_name = 'sample'
)
```

```python
vector_store.add_documents(pages)
```

```python
vector_store.get(include = ['embeddings','documents'])
```

```python
vector_store.similarity_search(
    query = 'what is the book about',
    k = 1
)
```

```python
vector_store.similarity_search_with_score(
    query= 'waht is the name of this book',
    # filter = {'author' : 'name of the book'}
)
```
