```python
import chromadb
from langchain_chroma import Chroma
from langchain_ollama import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter

PATH = r"C:\Users\Krish\Downloads\LangChain\RAG\VectorStores\chroma_db"

client = chromadb.PersistentClient(path=PATH)
for c in client.list_collections():
    print(c.name, c.count())          # find the one with count > 0
```

```python
vectorstore = Chroma(
    persist_directory=PATH,
    collection_name="sample",
    embedding_function=OllamaEmbeddings(model="nomic-embed-text"),
)
print(vectorstore._collection.count())
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 4})
docs = retriever.invoke("who is the author")
docs
```

```python
chunks = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=100).split_documents(docs)
```

```python
vs = Chroma.from_documents(chunks, OllamaEmbeddings    (model="nomic-embed-text"),
persist_directory=PATH, collection_name="sample")
print(vs._collection.count())   # must be > 0
```
