```python
from langchain_community.retrievers import WikipediaRetriever
```

```python
retriever = WikipediaRetriever(top_k_results= 2, lang = 'en', doc_content_chars_max=1000)
```

```python
queries = ["China-Pakistan relations", "Sino-Indian War", "India-Pakistan relations", "China-India relations"]
docs = [d for q in queries for d in retriever.invoke(q)]
```

```python
for i,doc in enumerate(docs):
    print(f'\n---Result{i+1} ---')
    print(f'content:\n{doc.page_content}...')
```
