```python
from langchain_classic.retrievers import MultiQueryRetriever
from langchain_ollama import OllamaEmbeddings
from langchain_core.documents import Document
from langchain_ollama.chat_models import ChatOllama
from langchain_community.vectorstores import FAISS
```

```python
all_docs = [
    Document(page_content="Walking regularly supports heart health and helps lower signs of depression.", metadata={"source": "H1"}),
    Document(page_content="Eating fruits and leafy greens aids bodily detox and supports a longer life.", metadata={"source": "H2"}),
    Document(page_content="Getting deep sleep is key for repairing cells and managing emotions.", metadata={"source": "H3"}),
    Document(page_content="Controlled breathing and mindfulness lower stress hormones and boost focus.", metadata={"source": "H4"}),
    Document(page_content="Staying well hydrated across the day supports energy levels and metabolism.", metadata={"source": "H5"}),
    Document(page_content="Home solar power setups help even out household electricity demand.", metadata={"source": "I1"}),
    Document(page_content="Python combines simplicity and capability, making it common in system design.", metadata={"source": "I2"}),
    Document(page_content="Plants generate energy from sunlight through the process of photosynthesis.", metadata={"source": "I3"}),
    Document(page_content="Qatar hosted the 2022 FIFA World Cup, sparking worldwide excitement.", metadata={"source": "I4"}),
    Document(page_content="Black holes warp spacetime and contain vast gravitational energy.", metadata={"source": "I5"}),
]
```

```python
embeddings = OllamaEmbeddings(model = 'nomic-embed-text')

vectorstore = FAISS.from_documents(documents = all_docs, embedding = embeddings)
```

```python
## Normal Retriver
similarity_retriver = vectorstore.as_retriever(search_type = 'similarity', kwargs= {'k' : 5})

## Multi Query Retriever
multy_query = MultiQueryRetriever.from_llm(
    retriever= vectorstore.as_retriever(kwargs= {'k' : 5}),
    llm= ChatOllama(model = 'llama3.2:3b')
)
```

```python
## Query
query = "How to imporve energy levels and maintain balance"
```

```python
similarity_results =  similarity_retriver.invoke(query)
multy_query_results  = multy_query.invoke(query)
```

```python
for i,docs in enumerate(similarity_results):
    print(f"{i+1} : {docs.page_content}")
```

```python
for i, docs in enumerate(multy_query_results):
    print(f"{i+1} : {docs.page_content}")
```
